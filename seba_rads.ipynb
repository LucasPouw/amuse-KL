```python
from amuse.community.seba.interface import SeBa
from amuse.lab import Particles, units

from make_system import SystemMaker
```

```python
seba = SeBa()

bodies = Particles(2)
bodies.mass = [2.8, 0.73] | units.MSun


seba.particles.add_particles(bodies)
seba.evolve_model(2.7 | units.Myr)
```

```python
print(bodies)
print(seba.particles.mass,seba.particles.radius)
seba.stop()
```

```python
seba2 = SeBa()
seba2.particles.add_particles(bodies)
seba2.evolve_model(100 | units.Myr)

print(seba2.particles.mass, seba2.particles.radius)
seba2.stop()
```
